```python
df["month"] = df["sale_date"].dt.to_period("M")
```

```python
monthly_sales = df.groupby("month")["total_sale"].sum()

print(monthly_sales)
```

```python
plt.figure(figsize=(14,6))

monthly_sales.plot(
    kind="line",
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
highest_sales_month = monthly_sales.idxmax()
highest_sales_value = monthly_sales.max()

print("Highest Sales Month:", highest_sales_month)
print(f"Highest Sales: ₹{highest_sales_value:,.2f}")
```

```python
lowest_sales_month = monthly_sales.idxmin()
lowest_sales_value = monthly_sales.min()

print("Lowest Sales Month:", lowest_sales_month)
print(f"Lowest Sales: ₹{lowest_sales_value:,.2f}")
```

```python
numeric_data = df.select_dtypes(include=np.number)

correlation = numeric_data.corr()

print(correlation)
```

```python
plt.figure(figsize=(11,8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Numerical Variables")

plt.tight_layout()
plt.show()
```

```python
corr_pairs = correlation.unstack()

corr_pairs = corr_pairs[
    corr_pairs.index.get_level_values(0) !=
    corr_pairs.index.get_level_values(1)
]

corr_pairs = corr_pairs.abs().sort_values(ascending=False)

print(corr_pairs.head(10))
```
